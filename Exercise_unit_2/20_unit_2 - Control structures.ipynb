### 1.3 Variable names

A **variable** stores a value. It is created the moment a value is assigned to it
with ``=``:

```python

```

Rules and conventions for the name:

* it must not start with a number and must not contain special characters (``-``, spaces, ...)
* it must not be a keyword of the language (``class``, ``if``, ``for``, ``True``, ...)
* variables are written in ``snake_case``: lower case, words separated by ``_``
* constants are written in ``ALL_CAPS``, class names in ``PascalCase``
* booleans sound like a question: ``is_active``, ``has_ticket``

### 1.4 Variables and their data type

Every value in Python has a **data type**. The most important ones are ``int``
(whole number), ``float`` (decimal number), ``str`` (text) and ``bool``
(``True`` / ``False``). The function ``type()`` tells you which one a variable has:

```python
x = 5
print(type(x))      # <class 'int'>
```

A variable can be overwritten with a value of a different type at any time.

### 1.5 Swapping two variables

An assignment **overwrites** the old value of a variable, so the order of your
statements matters. Whoever starts with ``a = b`` has lost the original value of
``a`` for good:

```python
a = 1
b = 2
a = b    # a is 2 now, the 1 is gone
b = a    # b is 2 as well - not what we wanted
```

**Task:** swap the values of the two variables below, so that ``a`` contains the
value of ``b`` and ``b`` the value of ``a``. Print both variables afterwards.
> Tip: you need a third (temporary) variable to store one value while overwriting it.

````python
a = "left"
b = "right"
````
