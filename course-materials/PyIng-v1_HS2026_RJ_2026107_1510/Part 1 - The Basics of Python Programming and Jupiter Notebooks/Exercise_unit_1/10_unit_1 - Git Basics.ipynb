# Exercise 1 — Git: solve the Markdown exercise on a branch and merge it back

## What you will do

You already know **Exercise 1 (Markdown)** — the notebook `12_unit_1 - Markdown.ipynb` in this
folder. This time you solve it **inside a git repository**, and you drive git **entirely from the
Visual Studio Code interface** — you never type a git command.

The workflow is the one real teams use:

1. put the untouched exercise into your repository and make a first commit on `main`;
2. create a **branch** for your solution;
3. **solve the Markdown exercise on that branch**, committing after every task;
4. **merge** the branch back into `main`.

**What you hand in:** your repository, with this history visible in **Git Graph** — a first
commit, a branch with several solution commits, and a merge that brings them back together.

## Before you start (from the platform)

## Step 2 — Create your solution branch

This step is not strictly necessary for git to work — you could always commit every change
straight to `main`. But what if you want to build something bigger, like a feature, and keep it
out of `main` until it is finished and tested? On `main` that is impossible, and it goes against
the whole idea of agile software development. This is what **branches** are for: a branch splits
off from `main`, so you can develop independently without touching `main`, and once you are sure
everything works and is ready to ship you merge all the changes back into `main`.

1. In the Source Control panel, click the **`⋯`** menu at the top → **Branch** →
   **Create Branch…**:

   ![The three-dot menu opened on Branch, with Create Branch highlighted](images/vs-code-git-panel.png)

2. Type the branch name **`solution`** and press Enter.
3. VS Code switches to the new branch automatically — the branch name in the blue status bar
   (bottom-left) now reads **`solution`**. `main` stays where it is.

*(Shortcut: clicking the branch name in the status bar also offers “Create new branch…”.)*

## Step 3 — Solve the Markdown tasks on the branch, one commit per task

Open `12_unit_1 - Markdown.ipynb` and work through tasks **1.1 – 1.7**.

**After each task, commit** (type a message → **✓ Commit** → **Yes**). Use a clear message each
time, for example:

* `Solve 1.1 headers`
* `Solve 1.2 text formatting`
* `Solve 1.3 table`
* `Solve 1.4 formula`
* `Solve 1.5 links and images`
* `Solve 1.6 lists`
* `Solve 1.7 quote`

Notes:

* Task **1.3** needs the *Nice Tables* extension (install it the same way as Git Graph).
* Task **1.5.2** asks you to create a `Pictures` folder and add an image — that image shows up in
  the Source Control panel too, so it gets committed with the next commit.

Aim for **at least four commits** on this branch, so the history is worth looking at in Git
Graph.

## Step 4 — Solve the code-cell tasks 2.1 – 2.2

Do tasks **2.1** (hello world) and **2.2** (arithmetic with `a` and `b`) in the same notebook,
then **commit** them (message `Solve 2.1 - 2.2 code cells`).

## Step 5 — See how the branch has diverged

1. Open **Git Graph**. The `solution` branch is now several commits ahead of `main`.
2. Switch to **`main`**: click the branch name in the status bar and pick `main` (or **`⋯` menu →
   Checkout to…**). Open the notebook — your answers are **gone** (they live safely on
   `solution`).
3. Switch back to **`solution`** the same way — your answers are back.

*In one sentence: what did the branch protect `main` from while you were solving?*

## Step 6 — Make `main` move too

So that the merge in the next step becomes a **real merge** (not just a shortcut), give `main`
its own commit first:

1. Switch to **`main`** (status bar, or **`⋯` menu → Checkout to…**).
2. Create a file **`README.md`** with your name and one line, e.g. *"My solutions to Exercise 1"*.
3. **Commit** it (message `Add README`).

Now **both** branches have a commit the other one does not have — open **Git Graph** and you will
see the line **fork into two**.

## Step 7 — Merge the solution back into `main`

You always merge *into* the branch you are currently on, so `main` has to be the current branch.

1. Switch to **`main`** (status bar, or **`⋯` menu → Checkout to…**). The status bar must read
   `main`.
2. In the Source Control panel, open the **`⋯`** menu → **Branch** → **Merge…** — the same
   submenu you used in Step 2:

   ![The three-dot menu opened on Branch, with Merge available](images/vs-code-git-panel.png)

3. Pick **`solution`** from the list. Keep the suggested commit message and confirm.
4. Git creates a **merge commit**. In **Git Graph** the two lines **join** back into one.
5. Optional: **`⋯` menu → Branch → Delete Branch… → `solution`** — the work is now in `main`.

> If you had **skipped Step 6**, `main` would not have moved and git would just **fast-forward**:
> no merge commit, the graph stays one straight line. Worth trying once to see the difference.

## Appendix — the same clicks as terminal commands (reference only)

You did everything with the GUI. For reference, these are the commands the GUI ran for you:

| GUI action | Command it runs |
|---|---|
| **Initialize Repository** | `git init` |
| type a message + **✓ Commit** → **Yes** | `git add -A` + `git commit -m "message"` |
| **`⋯` → Branch → Create Branch…** | `git switch -c <name>` |
| **`⋯` → Checkout to…** | `git switch <name>` |
| **`⋯` → Branch → Merge…** | `git merge <name>` |
| **`⋯` → Branch → Delete Branch…** | `git branch -d <name>` |
| **`⋯` → Push** | `git push` |
| the Git Graph window | `git log --oneline --graph --all` |
