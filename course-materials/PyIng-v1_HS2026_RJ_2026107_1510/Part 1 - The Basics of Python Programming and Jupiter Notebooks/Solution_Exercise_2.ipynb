### 1.4 Variables and their data type

Every value in Python has a **data type**. The most important ones are ``int``
(whole number), ``float`` (decimal number), ``str`` (text) and ``bool``
(``True`` / ``False``). The function ``type()`` tells you which one a variable has:

### 1.5 Swapping two variables

An assignment **overwrites** the old value of a variable, so the order of your
statements matters. Whoever starts with ``a = b`` has lost the original value of
``a`` for good:

```python
a = 1
b = 2
a = b    # a is 2 now, the 1 is gone
b = a    # b is 2 as well - not what we wanted
```

**Task:** swap the values of the two variables below, so that ``a`` contains the
value of ``b`` and ``b`` the value of ``a``. Print both variables afterwards.
> Tip: you need a third (temporary) variable to store one value while overwriting it.

````python
a = "left"
b = "right"

# Store the value of a in a temporary variable, otherwise it would be lost
temp = a
# Now a can be overwritten with the value of b
a = b
# b gets the original value of a back from the temporary variable
b = temp

print(a)    # right
print(b)    # left

# Python also offers a short form that does the same in one line:
# a, b = b, a
````

### 2.4 More arithmetic operators
A measurement took 3725 seconds. Print, by using ``//`` and ``%``:

* the full hours
* the remaining full minutes
* the remaining seconds

Print the result of ``2 ** 10`` as well.
> Tip: ``//`` divides and cuts off the decimals, ``%`` returns the remainder of a division. You need both twice.

````python
seconds = 3725

# One hour has 3600 seconds.
# // gives the full hours, % gives the seconds that are left over.
hours = seconds // 3600              # 3725 // 3600 = 1
remaining_seconds = seconds % 3600   # 3725 % 3600 = 125

# One minute has 60 seconds - the same trick a second time
minutes = remaining_seconds // 60    # 125 // 60 = 2
remaining_seconds = remaining_seconds % 60   # 125 % 60 = 5

print("hours:", hours)
print("minutes:", minutes)
print("seconds:", remaining_seconds)

# ** is the power operator: 2 to the power of 10
print(2 ** 10)     # 1024
````

### 2.5 Assignment operators

1. increase it by ``step``
2. decrease it by 5
3. double it
4. divide it by 4

At the end print the data type of ``counter`` and the value again as a whole number.
> Tip: ``counter += step`` is the short form of ``counter = counter + step``; ``int(value)`` cuts the decimals off.

````python
counter = 100
step = 7

counter += step     # same as counter = counter + step  -> 107
print(counter)

counter -= 5        # same as counter = counter - 5     -> 102
print(counter)

counter *= 2        # same as counter = counter * 2     -> 204
print(counter)

counter /= 4        # same as counter = counter / 4     -> 51.0
print(counter)

# Because of the division, counter is a float now
print(type(counter))    # <class 'float'>

# int() cuts the decimals off
print(int(counter))     # 51
````

### 2.11 Logical operators
Given are the three variables below. Print the result (``True`` or ``False``) of the following statements **without** using an ``if``:

* the person is at least 18 **and** has a ticket
* the person is younger than 6 **or** older than 65
* the person does **not** have a ticket
* the person may enter: at least 18 **and** (has a ticket **or** is a member)
* the opposite of the statement above — with ``not`` and brackets

````python
age = 20
has_ticket = False
is_member = True

# and: both sides have to be True
print(age >= 18 and has_ticket)                    # False

# or: one side being True is enough
print(age < 6 or age > 65)                         # False

# not: turns True into False and the other way round
print(not has_ticket)                              # True

# The brackets make sure that "or" is checked first
may_enter = age >= 18 and (has_ticket or is_member)
print(may_enter)                                   # True

# not in front of the whole bracket returns the opposite
print(not (age >= 18 and (has_ticket or is_member)))   # False
````

### 3.4 Leap year check
Write a program that checks whether a year is a leap year. A year is a leap year if it is dividable by 4, but not by 100, unless it is also dividable by 400.

````python
year = input("What year should be checked")
````

````python
# input() returns a string -> cast it to int to be able to calculate
year = int(year)

# Leap year rule:
# - dividable by 4 and NOT dividable by 100   -> leap year (e.g. 2024)
# - OR dividable by 400                        -> leap year (e.g. 2000)
# - everything else                            -> no leap year (e.g. 1900, 2023)
if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
````
